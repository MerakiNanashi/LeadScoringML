### Goal:

Validate the hypothesis that:


_Demonstrate that repeated optimization for observed conversion probability causes the ranking system to increasingly prioritize high-organic-conversion leads, resulting in reduced exposure diversity and suppression of intervention-sensitive leads over time._


The experiments should validate:

1. The existence of the feedback loop
2. The mismatch between conversion optimization and intervention value
3. Failure modes of naive top-K ranking
4. Effectiveness of reranking + exploration strategies
5. Practical viability of segment-level baseline correction
6. Long-term system behavior under repeated retraining

Questions:

1. Does pure conversion optimization collapse toward high-organic-conversion leads?

Show:
- exposure concentration
- reduced diversity
- suppression of uncertain leads

2. Are high-conversion leads always high-value intervention targets?

Show:
- high-organic leads may have low incremental lift
- mid-tier leads may have higher intervention value

3. Can reranking improve intervention efficiency?

Show:
- reranker improves total incremental conversions
- without catastrophic drop in overall conversion

4. Does increasing candidate pool size solve the issue?

Show:
- recall improves
- but selection bias persists

5. Can lightweight baseline correction approximate uplift-aware ranking?

Show:
- segment-baseline correction improves intervention allocation
- without requiring causal infrastructure

6. What are the long-term effects of repeated retraining?

Show:
- naive systems collapse
- exploration-aware systems maintain broader coverage
- reranking changes data distribution over time
